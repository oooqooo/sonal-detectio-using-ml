```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("Copy of sonar data.csv",header=None)
```

```python
df.head()
df.isnull().sum()
```

```python

# Step 1: Separate classes
rock = df[df[60] == "R"]
mine = df[df[60] == "M"]

# Step 2: Take samples
rock_sample = rock.iloc[:97, :]
mine_sample = mine.iloc[:97, :]

print(rock_sample.shape)
print(mine_sample.shape)

# Step 3: Combine
df_new = pd.concat((rock_sample, mine_sample), axis=0)

# Step 4: Define X and y
X = df_new.drop(columns=60, axis=1)
y = df_new[60]

# Step 5: Split
x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, stratify=y, random_state=42
)

# Step 6: Scaling
scaler = StandardScaler()
x_train_std = scaler.fit_transform(x_train)
x_test_std = scaler.transform(x_test)

# Step 7: Train model
model = LogisticRegression(solver='liblinear', max_iter=1000,C=0.1)
model.fit(x_train_std, y_train)

# Step 8: Test accuracy (IMPORTANT)
pred = model.predict(x_test_std)
acc = accuracy_score(y_test, pred)
print("Accuracy:", acc)
pred = model.predict(x_train_std)
acc = accuracy_score(y_train, pred)

print("Accuracy:", acc)
```
